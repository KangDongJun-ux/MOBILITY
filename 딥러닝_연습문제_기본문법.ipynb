## **concat, stack($\star$)**


a,b가 아래와 같이 주어졌다고 하자.
```python
a = torch.tensor([1]*10)
b = torch.tensor([2]*10)
```
아래를 잘 읽고 물음에 답하라.

`-` (1) 주어진 a,b와 torch.concat를 이용하여 아래와 같은 배열을 만들어라.

````python
a = torch.tensor([1]*10)
b = torch.tensor([2]*10)
a,b
````

````python
a.shape, b.shape
````

````python
torch.concat([a.reshape(1,-1), b.reshape(1,-1)], axis=1)
````

`-` (2) 주어진 a,b 와 torch.concat,.reshape를 이용하여 아래와 같은 배열을 만들어라.

````python
torch.concat([a.reshape(-1,1), b.reshape(-1,1)])
````

`-` (3) 주어진 a,b 와 torch.concat,.reshape를 이용하여 아래와 같은 배열을 만들어라.

````python
torch.concat([a.reshape(-1,1), b.reshape(-1,1)],axis=1)
````

`-` (4) 주어진 a,b와 torch.stack 을 이용하여 아래와 같은 배열을 만들어라.

````python
torch.stack([a,b],axis=0)
````

`-` (5) 주어진 a,b와 torch.stack을 이용하여 아래와 같은 배열을 만들어라.

````python
torch.stack([a,b],axis=1) # stack은 벡터로 처리가 되는 데 축 방향만 설정해주면 열 벡터로도 계산이 가능한 것을 보임.
````

## **torch.einsum**

`-` (1) 아래에 코드중 X.t()에 대응하는 코드를 torch.einsum으로 구현하라.

```python
X = torch.randn(5,2)
X.t()
```
tensor([[ 0.2055,  0.0693,  0.2733, -0.0948, -0.9798],

[ 0.6524, -0.0180,  0.1093, -2.3999,  0.5101]])


````python
X = torch.randn(5,2)
X.t()
````

````python
torch.einsum("ij ->ji", X)
````

`-` (2) 아래에 코드중 X@b에 대응하는 코드를 torch.einsum으로 구현하라.
```python
X = torch.randn(5,2)
b = torch.randn(2,1)
X@b
```


````python
X = torch.randn(5,2)
b = torch.randn(2,1)
X@b
````

````python
torch.einsum("ij,jk ->ik", X,b)
````

`-` (3) 아래에 코드중 linr(X)에 대응하는 코드를 torch.einsum으로 구현하라.
```python
X =  torch.randn(5,2)
linr = torch.nn.Linear(2,1,bias=False)
linr(X)
```


````python
X =  torch.randn(5,2)
linr = torch.nn.Linear(2,1,bias=False)
linr(X)
````

````python
torch.einsum("ij,jk->ik", X,linr.weight.data.T)
````
