## A. Torch

```python
import torch
```

`-` 벡터

```python
torch.tensor([1,2,3]) # 길이가 3인 벡터
```

`-` 벡터의 덧셈

```python
torch.tensor([1,2,3]) + torch.tensor([2,2,2]) # 각 원소끼리 더함.
```

`-` 브로드캐스팅

```python
torch.tensor([1,2,3]) + 2 # 원래는 벡터끼리만 더하기가 가능하지만 이렇게도 허용함. 2 --> torch.tensor([2,2,2]) 처럼 작동하는 것으로 보임.
```

##  B. 벡터와 매트릭스

`-` 3 X 2 matrix

```python
torch.tensor([[1,2],[3,4],[5,6]]) # [1.2] 하나가 행, [1.2]에 두개의 열이 있음. 즉 이 메트릭스는 3 X 2 행렬임.
```

`-` 3 X 1 matrix = 3 X 1 column vector

```python
torch.tensor([[1],[2],[3]]) # 열 벡터
```

`-` 1 X 2 matrix = 1 X 2 row vertor

```python
torch.tensor([[1,2]]) # 행 벡터
```

`-` 더하기

**브로드캐스팅(편한거)**

```python
torch.tensor([[1,2],[3,4],[5,6]]) - 1 # 1이 torch.tensor([[1,1],[1,1],[1,1]]) 과 같이 작동함.
```

```python
torch.tensor([[1,2],[3,4],[5,6]]) + torch.tensor([[-1],[-3],[-5]])  # 뒷 항이 torch.tensor([[-1.-1],[-1,-1],[-1,-1]])으로 작동함. 즉 나머지 열을 채움
```

```python
torch.tensor([[1,2],[3,4],[5,6]]) + torch.tensor([[-1,-2]]) # 뒷 항이 torch.tensor([[-1,-2],[-1,-2],[-1,-2]])처럼 작동함.  즉 나머지 행을 채움
```

**잘못된 브로드캐스팅**

```python
torch.tensor([[1,2],[3,4],[5,6]]) + torch.tensor([[-1,-3,-5]]) # 행이나 열의 개수가 하나라도 같아야 브로드캐스팅이 되는 듯??
```

```python
torch.tensor([[1,2],[3,4],[5,6]]) + torch.tensor([[-1],[-2]]) # 행이나 열의 개수가 하나도 안맞으면 브로드캐스팅 안되는 듯??
```

**이상한 것**

```python
torch.tensor([[1,2],[3,4],[5,6]]) + torch.tensor([-1,-2]) # 3 X 2 matrix 와 크기가 2인 vector의 합이 가능   # 벡터 항이 torch.tensor([[-1,-2]])처럼 작동해서 브로드캐스팅이 가능한 듯??
```

```python
torch.tensor([[1,2],[3,4],[5,6]]) + torch.tensor([-1,-3,-5])
```

`-` 행렬곱

**정상적인 행렬곱**

```python
torch.tensor([[1,2],[3,4],[5,6]]) @ torch.tensor([[1],[2]]) #  3 X 2 @ 2 X 1 = 3 X 1
```

```python
torch.tensor([[1,2,3]]) @ torch.tensor([[1,2],[3,4],[5,6]]) # 1 X 3 @ 3 @ 2 = 1@ 2
```

**잘모된 행렬곱**

```python
torch.tensor([[1,2],[3,4],[5,6]]) @ torch.tensor([[1,2]]) # 3 X 2 @ 1 X 2 = X
```

```python
torch.tensor([[1],[2],[3]]) @ torch.tensor([[1,2],[3,4],[5,6]]) # 3 X 1 @ 3 X 2 = X
```

**이상한 것**

```python
torch.tensor([[1,2],[3,4],[5,6]]) @ torch.tensor([1,2]) # 행렬과 벡터 곱인데 가능 --> 결과 : 길이가 3인 벡터
                                                                                      # 결과:  각 원소는 각 행과 [1,2]의 내적
                                                                                      # [1,2] * [1,2] = 1*1 + 2*2 = 5
```

```python
torch.tensor([1,2,3]) @ torch.tensor([[1,2],[3,4],[5,6]]) # 결과: 길이가 2인 벡터
                                                                                         # 결과: 각 원소는 각 열과 [1,2,3]의 내적
                                                                                         # [1.2.3] * [1,3,5] = 1*1 + 2*3  + 3*5 = 22
```

## C. transpose, reshape

`-` transpose

```python
torch.tensor([[1,2],[3,4]]).T # 열과 행을 바꿈 대각선을 기준으로 바꿈
```

```python
torch.tensor([[1], [3]]).T # 2 X 1 matrix --> 1 X 2 matrix
```

```python
torch.tensor([[1,2]]).T # 1 X 2 matrix --> 2 X 1 matrix
```

`-` reshape

**일반적인 사용**

```python
torch.tensor([[1,2],[3,4],[5,6]]).reshape(2,3)
```

```python
torch.tensor([[1,2],[3,4],[5,6]])
```

```python
torch.tensor([[1,2],[3,4],[5,6]]).reshape(1,6)
```

```python
torch.tensor([[1,2],[3,4],[5,6]]).reshape(6)  # 길이가 6인 벡터
```

**편한 것**

```python
torch.tensor([[1,2],[3,4],[5,6]]).reshape(2,-1) # 2행으로 맞추고 싶을 때 뒷 열을 -1로 넣어주면 자동으로 열을 맞춰줌.
```

```python
torch.tensor([[1,2],[3,4],[5,6]]).reshape(-1) # 1차원으로 변환(평탄)
```

## D. concat, stack $(\star\star\star)$

`-` concat

```python
a = torch.tensor([[1],[3],[5]]) # 3 X 1 matrix
b = torch.tensor([[2],[4],[6]]) # 3 X 1 matrix
torch.concat([a,b], axis = 1) # 3 X 2 matrix  # axis를 지정해서 열로 붙을지 행으로 붙을 지 정할 수 있다.
# concat 은 matrix 기준 가능
```

`-` stack

```python
a = torch.tensor([1,3,5])
b = torch.tensor([2,4,6])
torch.stack([a,b], axis = 1 )
# stack 은 벡터 기준에서 가능
```
